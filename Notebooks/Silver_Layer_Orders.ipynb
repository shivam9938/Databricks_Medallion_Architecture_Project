## **Data Reading**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import *

df = spark.read.format("parquet").load("abfss://bronze@dbproject.dfs.core.windows.net/orders")
df.display()
```

```python
df.groupBy("order_id").count().filter("count > 1").display()
```

## **Data Transformation**

```python
df = df.withColumn("unit_price", round(col("total_amount")/col("quantity"),2))
df.display()
```

```python
df = df.drop('_rescued_data')
df.display()
```

```python
customer_window = Window.partitionBy("customer_id")

df = df.withColumn(
    "customer_avg_order_value",
    round(avg("total_amount").over(customer_window),2)
)
df.display()
```

```python
customer_window = Window.partitionBy("customer_id")

df = df.withColumn(
    "customer_total_spend",
    sum(col("total_amount")).over(customer_window)
)

df = df.withColumn("ptg_of_customer_spend_on_order", round(col("total_amount") / col("customer_total_spend")*100, 2))
df = df.drop("customer_total_spend")
df.display()
```

## **Data Writing**

```python
df.write.format("delta").mode("overwrite").save("abfss://silver@dbproject.dfs.core.windows.net/orders")
```

```python
%sql
CREATE TABLE IF NOT EXISTS db_catalog.silver.orders_silver
USING DELTA
LOCATION 'abfss://silver@dbproject.dfs.core.windows.net/orders'
```
